2. Exact video-generation prompt
Create a 3-second, 16:9 video using all three uploaded Exercise 2 images as visual references. Preserve the important subjects, colours, and visual identity of each reference. Begin with the virtual AI training session: show the woman trainer presenting confidently while the digital AI graphics glow subtly and the participant video grid remains visible. Then create a smooth, fluid transition in which the training-screen imagery transforms into the playful mouse-and-cat garden chase. The small brown mouse runs quickly across the green lawn while the grey-and-white cat follows closely behind; use lively, exaggerated cartoon movement without changing the characters' appearance. Finish by transitioning into the supplied blended third image, making it the clear final visual. The transition should feel creative and continuous, as if the subjects and visual elements from the first two references are merging into the blended composition. Maintain clean details, stable subjects, natural motion, and consistent lighting. No extra characters, no new text, no logos, no flicker, no abrupt cuts, and no distorted faces or objects.

3. Camera and transition instructions
- Duration / format: 3 seconds, 16:9 widescreen.
- 0.0–0.9 seconds: Start with Image 1. Use a gentle camera push-in toward the trainer and presentation screen; keep the participant grid readable.
- 0.9–2.0 seconds: Use a smooth morph/dissolve transition into Image 2. Follow the mouse with a subtle left-to-right tracking movement; keep the cat visible behind it.
- 2.0–3.0 seconds: Blend the chase scene into Image 3 with a soft, continuous morph. Settle on the blended image with a brief, steady final frame.
- Camera style: Smooth, controlled movement; no sudden zooms, shakes, or rapid cuts.
- Visual consistency: Retain the reference images' main subjects, colour palette, and overall style. Avoid introducing unrelated elements.
