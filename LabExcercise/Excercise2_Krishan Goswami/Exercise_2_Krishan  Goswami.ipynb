#### Exercise 2: Computer Vision - Image Blending Using OpenCV

##### Import required libraries

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

##### Load two sample images

```python
# Load two input images
img1 = cv2.imread("pic.jpeg")
img2 = cv2.imread("home1.png")

# Check whether images loaded successfully
if img1 is None or img2 is None:
    raise FileNotFoundError("Please check image1.jpg and image2.jpg")

print("Image 1 Shape:", img1.shape)
print("Image 2 Shape:", img2.shape)
```

##### Resize both images

```python
# Set common dimensions
width = 500
height = 400

img1 = cv2.resize(img1, (width, height))
img2 = cv2.resize(img2, (width, height))

print("Resized Image 1:", img1.shape)
print("Resized Image 2:", img2.shape)
```

##### Perform weighted image blending

```python
# Set blending weights
alpha = 0.6
beta = 0.4
gamma = 0

# Blend images
blended_img = cv2.addWeighted(
    img1, alpha,
    img2, beta,
    gamma
)

print("Image blending completed successfully!")
```

##### Display original and blended images

```python
# Convert BGR to RGB for Matplotlib
img1_rgb = cv2.cvtColor(img1, cv2.COLOR_BGR2RGB)
img2_rgb = cv2.cvtColor(img2, cv2.COLOR_BGR2RGB)
blended_rgb = cv2.cvtColor(blended_img, cv2.COLOR_BGR2RGB)

# Display all three images
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(img1_rgb)
plt.title("Original Image 1")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(img2_rgb)
plt.title("Original Image 2")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(blended_rgb)
plt.title("Blended Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```

##### Save the images

```python
# Save resized input images
cv2.imwrite("Exe2_inp1_kri.jpg", img1)
cv2.imwrite("Exe2_inp2_kri.jpg", img2)

# Save blended image
cv2.imwrite("Exe2_Blend_kri.jpg", blended_img)

# Save comparison containing all three images
comparison = np.hstack((img1, img2, blended_img))

cv2.imwrite("Exe2_Comp_kri.jpg", comparison)

print("All images saved successfully!")
```

##### Verify the saved blended image

```python
# Load saved blended image
saved_img = cv2.imread("Exe2_Blend_kri.jpg.jpg")

# Verify dimensions
print("Saved Image Shape:", saved_img.shape)

# Display saved image
plt.figure(figsize=(7, 5))

plt.imshow(cv2.cvtColor(saved_img, cv2.COLOR_BGR2RGB))
plt.title("Saved Blended Image")
plt.axis("off")

plt.show()
```
