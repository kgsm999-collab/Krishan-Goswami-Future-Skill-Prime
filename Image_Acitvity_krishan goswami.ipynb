```python

import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread(r"C:\Users\HLBS\Desktop\pic.jpeg")

print(type(img))
```

```python
print(img.shape)
```

```python
# Check if the image successfully loaded
if img is None:
    print("Error: Could not load image. Check the file path and name!")
else:
    # First argument must be the window name string, second is the image array
    print("showing image")
    cv2.imshow('Bootcamp Image', img) 
    cv2.waitKey(0)

cv2.destroyAllWindows()
```

```python

img_resize = cv2.resize(img, (300, 200), interpolation=cv2.INTER_AREA)

print(f"Min pixel value: {img_resize.min()}, Max pixel value: {img_resize.max()}")
print(f"Shape of resized image: {img_resize.shape}")

cv2.imshow("Resized Image", img_resize)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python

print(img.dtype)

img_flip = cv2.flip(img,0)

cv2.imshow("Bootcamp",img_flip)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python



img_crop=img[100:300,200:500]

cv2.imshow("Bootcamp",img_crop)
cv2.waitKey(0)
cv2.destroyAllWindows()

cv2.imwrite('cropped_image.jpg', img_crop)
```
