```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Set seed for reproducibility
np.random.seed(42)

# Generate random numerical data
df = pd.DataFrame({
    'f1': np.random.randint(1, 100, 50),
    'f2': np.random.randint(1, 100, 50),
    'f3': np.random.randint(1, 100, 50),
    'f4': np.random.randint(1, 100, 50)
})

# Display first 5 rows
df.head()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df['f2'],
    df['f3'],
    color='blue',
    alpha=0.7
)

plt.xlabel("f2")
plt.ylabel("f3")
plt.title("Scatter Plot: f2 vs f3")
plt.grid(True, alpha=0.3)

plt.show()
```

```python
# Calculate mean across f1, f2, f3 and f4
df['f5'] = df[['f1', 'f2', 'f3', 'f4']].mean(axis=1)

# Display updated DataFrame
df.head()
```

```python
# Export DataFrame to an Excel file
file_name = "LabExercise1_Krishan Goswami.xlsx"

df.to_excel(
    file_name,
    index=False,
    sheet_name="EDA_Report"
)

print("Excel file generated successfully:", file_name)
```

```python
# Read the exported Excel file
exported_df = pd.read_excel("LabExercise1_Krishan Goswami.xlsx")

# Display the exported data
display(exported_df.head())

print("Shape:", exported_df.shape)
print("Columns:", exported_df.columns.tolist())
```
